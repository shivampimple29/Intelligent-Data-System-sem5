```python
#Exp3

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import sklearn as s

import kagglehub

import kagglehub

# Download latest version
path = kagglehub.dataset_download("uciml/iris")

print("Path to dataset files:", path)
```

```python
#Step 1 : Load the Dataset

iris = pd.read_csv("/kaggle/input/iris/Iris.csv")
df = pd.DataFrame(iris)

print("\n---------------------HEAD---------------------\n")
print(df)
```

```python
# used to show relation between 2 vars
# used to detect correlation between 2 vars
# strong pattern = strong correlation
# random points = weak or no correlation

plt.scatter(df['PetalLengthCm'], df['PetalWidthCm'])
plt.title('Petal Length vs Petal Width')
plt.xlabel('PetalLengthCm (cm)')
plt.ylabel('PetalWidthCm (cm)')
plt.show()
```

```python
plt.scatter(df['SepalLengthCm'], df['SepalWidthCm'])
plt.title('Sepal Length vs Sepal Width')
plt.xlabel('SepalLengthCm (cm)')
plt.ylabel('SepalWidthCm (cm)')
plt.show()
```

```python
sns.boxplot(x=df['SepalWidthCm'])
plt.title("Box Plot of Sepal Width")
plt.show()

# Analysis
```

```python
sns.boxplot(x=df['SepalLengthCm'])
plt.title("Box Plot of Sepal Length")
plt.show()

# Analysis
# median is around 5.7
# 50% of data is around
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Create subplots
fig, axes = plt.subplots(2, 2, figsize=(10, 8))

columns = ["SepalLengthCm", "SepalWidthCm", "PetalLengthCm", "PetalWidthCm"]

for ax, col in zip(axes.flat, columns):
    sns.boxplot(x=df[col], ax=ax)
    ax.set_title(f"Box Plot of {col}")

plt.tight_layout()
plt.show()
```

```python
# Plot 3: Multiple Box Plot - Petal Length by Species
sns.boxplot(data=df, x='Species', y='PetalLengthCm')
plt.title('Petal Length by Species')
plt.show()
```

```python
# Plot 3: Multiple Box Plot - Sepal Width by Species
sns.boxplot(data=df, x='Species', y='SepalWidthCm')
plt.title('Sepal Width by Species')
plt.show()
```

```python
ax = sns.countplot(
    data=df,
    x='Species',
    hue='Species',
    palette='Set2',
    legend=False
)
# Display count above each bar
for container in ax.containers:
    ax.bar_label(container)

plt.title('Distribution of Iris Species')
plt.show()
```

```python
correlation = df.corr(numeric_only=True)
correlation
```

```python
sns.heatmap(correlation, annot=True)
plt.title('Correlation Heatmap')
plt.show()
```

```python
sns.pairplot(df, hue='Species')

# Pair plot creates a grid of plots comparing every feature with every other feature.
# It helps visualize relationships between numerical attributes.

plt.show()
```

```python
# Plot 7: Histogram
# Step 18: Histogram (Histplot)

sns.histplot(data=df, x='SepalLengthCm', hue='Species', kde=True)

plt.title("Histogram of Sepal Length")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.hist(x=df['SepalLengthCm'], bins=10, color='skyblue', edgecolor='black')

plt.title('Histogram of Sepal Length')
plt.xlabel('Sepal Length')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(7,5))

sns.kdeplot(data=df, x='PetalLengthCm', hue='Species', fill=True)

plt.title('KDE Plot of Petal Length by Species')
plt.xlabel('Petal Length (cm)')
plt.ylabel('Density')

plt.show()
```

```python
plt.figure(figsize=(7,5))

sns.kdeplot(data=df, x='SepalLengthCm', hue='Species', fill=True)

plt.title('KDE Plot of Sepal Length by Species')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Density')

plt.show()
```

```python
# Plot 8: KDE Plot
# Step 19: Kernel Density Estimation (KDE) Plot

plt.figure(figsize=(6,4))

sns.kdeplot(data=df, x='PetalLengthCm', fill=True, color='purple')

plt.title('KDE Plot of Petal Length')
plt.xlabel('Petal Length (cm)')
plt.ylabel('Density')

plt.show()
```

```python
# Histogram for Sepal Length

plt.figure(figsize=(6,4))

sns.histplot(data=df, x='SepalLengthCm', kde=True)

plt.title('Histogram of Sepal Length')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Frequency')

plt.show()
```

```python
# Scatter Plot

plt.figure(figsize=(6,4))

plt.scatter(df['PetalLengthCm'], df['PetalWidthCm'])

plt.title('Petal Length vs Petal Width')
plt.xlabel('Petal Length (cm)')
plt.ylabel('Petal Width (cm)')

plt.show()
```
