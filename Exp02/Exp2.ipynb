```python
#part B

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import sklearn as s

#step 1 : load dataset
iris=pd.read_csv("/kaggle/input/iris-dataset/iris.csv")
df1=pd.DataFrame(iris)
print(df1)
print("\n---------------------HEAD-----------------------\n")
print(df1.head())
```

```python
df1.loc[2, 'sepal_length'] = 9
df1.loc[4, 'sepal_length'] = 10

# feature
feature=df1['sepal_length']

Q1=feature.quantile(0.25)
Q2=feature.quantile(0.50)
Q3=feature.quantile(0.75)
print("Q1 :",Q1, " Q2 :",Q2," Q3:",Q3)

#step 4

IQR=Q3 - Q1
print(IQR)

#step 5

lower=Q1-1.5*IQR
upper=Q3+1.5*IQR

print("lower:",lower," upper:",upper)

#step 6

outliers = df1[(feature < lower) | (feature > upper)]
print(outliers)

# Step 7: Plot Boxplot
plt.figure(figsize=(6,4))
plt.boxplot(feature, vert=True)

# Add horizontal lines for quartiles
# axhline() is used to draw a horizontal line across the plot
plt.axhline(Q1, color='blue', linestyle='--', label='Q1')
plt.axhline(Q2, color='green', linestyle='--', label='Median (Q2)')
plt.axhline(Q3, color='red', linestyle='--', label='Q3')

# Labels and title
plt.title("Box Plot with Q1, Q2, Q3 (Sepal Length)")
plt.ylabel("Sepal Length")
plt.legend()

plt.show()
```

```python
# -------------------------------
# Step 1: Import Required Libraries
# -------------------------------

import pandas as pd
from sklearn.preprocessing import StandardScaler

# -------------------------------
# Step 2: Create Some Outliers
# -------------------------------
# The original sepal_length values are approximately between 4 and 8.
# Here, we intentionally insert extreme values so that the Z-score
# method can identify them as outliers.

df1.loc[2, 'sepal_length'] = 12
df1.loc[4, 'sepal_length'] = 13

# -------------------------------
# Step 3: Select the Feature
# -------------------------------
# StandardScaler works on numerical data.
# We select only the 'sepal_length' column.

feature = df1[['sepal_length']]

# -------------------------------
# Step 4: Create StandardScaler Object
# -------------------------------
# StandardScaler standardizes the data.
#
# Formula:
# Z = (X - Mean) / Standard Deviation
#
# It converts every value into its Z-score.

scaler = StandardScaler()

# -------------------------------
# Step 5: Fit and Transform
# -------------------------------
# fit()       -> Calculates the Mean and Standard Deviation.
#
# transform() -> Converts every value into a Z-score.
#
# fit_transform() performs both operations together.

z_scores = scaler.fit_transform(feature)

# -------------------------------
# Step 6: Convert Output into DataFrame
# -------------------------------
# fit_transform() returns a NumPy array.
# Converting it into a DataFrame makes the output easier to read.

z_df = pd.DataFrame(z_scores, columns=['Z-Score'])

print("Z-Score of each observation:")
print(z_df)

# -------------------------------
# Step 7: Detect Outliers
# -------------------------------
# Generally,
#
# Z-score > 3
# or
# Z-score < -3
#
# indicates an outlier.

outliers = df1[(z_df['Z-Score'] > 3) | (z_df['Z-Score'] < -3)]

print("\nDetected Outliers:")
print(outliers)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# ---------------------------------
# Use Iris Dataset Feature
# ---------------------------------

feature = df1['sepal_length']


# Calculate mean and standard deviation

mean = feature.mean()
std = feature.std()

print("Mean:", mean)
print("Standard Deviation:", std)


# ---------------------------------
# Create Gaussian Curve
# ---------------------------------

x = np.linspace(
    feature.min()-3*std,
    feature.max()+3*std,
    1000
)

y = norm.pdf(x, mean, std)


# ---------------------------------
# Plot Curve
# ---------------------------------

plt.figure(figsize=(12,6))

plt.plot(
    x,
    y,
    color='black',
    linewidth=3
)


# ---------------------------------
# Shade Regions
# ---------------------------------

# Normal region (-1σ to +1σ)

plt.fill_between(
    x, y,
    where=(x >= mean-std) & (x <= mean+std),
    color='yellow',
    alpha=0.6,
    label='Normal (-1σ to +1σ)'
)


# Slightly unusual (-2σ to -1σ and +1σ to +2σ)

plt.fill_between(
    x, y,
    where=((x >= mean-2*std) & (x < mean-std)) |
          ((x > mean+std) & (x <= mean+2*std)),
    color='lightgreen',
    alpha=0.7,
    label='Slightly unusual (±1σ to ±2σ)'
)


# Moderately unusual (-3σ to -2σ and +2σ to +3σ)

plt.fill_between(
    x, y,
    where=((x >= mean-3*std) & (x < mean-2*std)) |
          ((x > mean+2*std) & (x <= mean+3*std)),
    color='orange',
    alpha=0.7,
    label='Moderately unusual (±2σ to ±3σ)'
)


# Outliers (< -3σ and > +3σ)

plt.fill_between(
    x, y,
    where=(x < mean-3*std) | (x > mean+3*std),
    color='red',
    alpha=0.7,
    label='Outliers (> ±3σ)'
)


# ---------------------------------
# Add Vertical Lines
# ---------------------------------

# Mean (0 sigma)

plt.axvline(
    mean,
    color='black',
    linestyle='--',
    linewidth=2,
    label='Mean (μ)'
)


# ±1 Sigma

plt.axvline(
    mean-std,
    color='green',
    linestyle='--',
    label='-1σ'
)

plt.axvline(
    mean+std,
    color='green',
    linestyle='--',
    label='+1σ'
)


# ±2 Sigma

plt.axvline(
    mean-2*std,
    color='orange',
    linestyle='--',
    label='-2σ'
)

plt.axvline(
    mean+2*std,
    color='orange',
    linestyle='--',
    label='+2σ'
)


# ±3 Sigma

plt.axvline(
    mean-3*std,
    color='red',
    linestyle='--',
    label='-3σ'
)

plt.axvline(
    mean+3*std,
    color='red',
    linestyle='--',
    label='+3σ'
)


# ---------------------------------
# Labels
# ---------------------------------

plt.title(
    "Gaussian Distribution of Iris Sepal Length\nwith Z-Score Regions"
)

plt.xlabel("Sepal Length")
plt.ylabel("Probability Density")

plt.legend(
    bbox_to_anchor=(1.05,1),
    loc='upper left'
)

plt.grid(alpha=0.3)

plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.preprocessing import StandardScaler

# ----------------------------------------
# Step 1: Create Sample Outliers
# ----------------------------------------

df1.loc[2, 'sepal_length'] = 12
df1.loc[4, 'sepal_length'] = 13

# ----------------------------------------
# Step 2: Select Feature
# ----------------------------------------

feature = df1[['sepal_length']]

# ----------------------------------------
# Step 3: Apply Z-Score Scaling
# ----------------------------------------
# StandardScaler standardizes the data using:
#
#           Z = (X - Mean) / Standard Deviation
#
# After scaling:
# Mean ≈ 0
# Standard Deviation ≈ 1

scaler = StandardScaler()

scaled_feature = scaler.fit_transform(feature)

# Convert to DataFrame for easier visualization
scaled_df = pd.DataFrame(
    scaled_feature,
    columns=['Scaled Sepal Length']
)

# ----------------------------------------
# Step 4: Plot Before and After Histograms
# ----------------------------------------

fig, ax = plt.subplots(1, 2, figsize=(14,5))

# BEFORE SCALING
ax[0].hist(feature['sepal_length'],
           bins=12,
           color='skyblue',
           edgecolor='black')

ax[0].set_title("Before Z-Score Scaling")
ax[0].set_xlabel("Sepal Length")
ax[0].set_ylabel("Frequency")

# AFTER SCALING
ax[1].hist(scaled_df['Scaled Sepal Length'],
           bins=12,
           color='lightgreen',
           edgecolor='black')

ax[1].set_title("After Z-Score Scaling")
ax[1].set_xlabel("Z-Score")
ax[1].set_ylabel("Frequency")

plt.tight_layout()
plt.show()
```
