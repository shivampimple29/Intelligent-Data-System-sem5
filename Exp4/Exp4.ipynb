# A* (A-Star) Search Algorithm

A* is a path-finding algorithm used to find the shortest path from a source node to a goal node.

It uses the following formula:

f(n) = g(n) + h(n)

where:
- g(n) = actual cost from the source node to node n
- h(n) = heuristic estimate from node n to the goal
- f(n) = total estimated cost

## Steps of A* Algorithm

1. Start from the source node.
2. Set g(start) = 0.
3. Calculate f(n) = g(n) + h(n).
4. Store the nodes in the Open List.
5. Select the node having the smallest f(n).
6. Expand its neighboring nodes.
7. Update the cost values if a better path is found.
8. Move the explored node to the Closed List.
9. Repeat the process until the goal node is reached.

```python
graph = {
    "A": [("B", 1), ("C", 4)],
    "B": [("D", 3), ("C", 2)],
    "C": [("E", 5)],
    "D": [("G", 4), ("F", 2)],
    "E": [("G", 3)],
    "F": [("G", 1)],
    "G": []
}

heuristic = {
    "A": 5,
    "B": 6,
    "C": 4,
    "D": 3,
    "E": 3,
    "F": 1,
    "G": 0
}
```

```python
def a_star_search(graph, heuristic, start_node, goal_node):

    # Step 1: Start from the source node

    # Step 2: Set g(start) = 0
    g_cost = {
        start_node: 0
    }

    # Step 3: Calculate f(start) = g(start) + h(start)
    f_cost = {
        start_node: g_cost[start_node] + heuristic[start_node]
    }

    # Step 4: Store the start node in the Open List
    open_list = [start_node]

    # Closed List
    closed_list = []

    # Store the path
    came_from = {
        start_node: None
    }

    while open_list:

        # Step 5: Select the node having the smallest f(n)
        current_node = min(
            open_list,
            key=lambda node: f_cost[node]
        )

        # Check whether goal is reached
        if current_node == goal_node:

            path = []

            while current_node is not None:
                path.append(current_node)
                current_node = came_from[current_node]

            path.reverse()

            return path, g_cost[goal_node]

        # Step 6: Expand neighboring nodes
        for neighbor_node, edge_cost in graph[current_node]:

            if neighbor_node in closed_list:
                continue

            # Calculate new g(n)
            new_g_cost = (
                g_cost[current_node] + edge_cost
            )

            # Step 7: Update cost if a better path is found
            if (
                neighbor_node not in g_cost
                or new_g_cost < g_cost[neighbor_node]
            ):

                g_cost[neighbor_node] = new_g_cost

                # Step 3: Calculate f(n) = g(n) + h(n)
                f_cost[neighbor_node] = (
                    g_cost[neighbor_node]
                    + heuristic[neighbor_node]
                )

                came_from[neighbor_node] = current_node

                # Step 4: Add node to Open List
                if neighbor_node not in open_list:
                    open_list.append(neighbor_node)

        # Step 8: Move explored node to Closed List
        open_list.remove(current_node)
        closed_list.append(current_node)

    # No path found
    return None, float("inf")
```

```python
start_node = "A"
goal_node = "G"

shortest_path, total_cost = a_star_search(
    graph,
    heuristic,
    start_node,
    goal_node
)

print("Shortest path:", " -> ".join(shortest_path))
print("Total cost:", total_cost)
```

```python
import heapq


def a_star_search(graph, heuristic, start_node, goal_node):
    """Find the lowest-cost path from start_node to goal_node using A*."""

    priority_queue = []

    # Store: (f(n), g(n), node)
    heapq.heappush(
        priority_queue,
        (heuristic[start_node], 0, start_node)
    )

    # Stores the previous node in the shortest path
    came_from = {
        start_node: None
    }

    # Stores the lowest known cost from start to each node
    cost_so_far = {
        start_node: 0
    }

    while priority_queue:

        _, current_cost, current_node = heapq.heappop(priority_queue)

        # Goal reached
        if current_node == goal_node:

            path = []

            while current_node is not None:
                path.append(current_node)
                current_node = came_from[current_node]

            path.reverse()

            return path, current_cost

        # Check all neighboring nodes
        for neighbor_node, edge_cost in graph.get(current_node, []):

            new_cost = (
                cost_so_far[current_node] + edge_cost
            )

            # If this path is better than the previous path
            if (
                neighbor_node not in cost_so_far
                or new_cost < cost_so_far[neighbor_node]
            ):

                cost_so_far[neighbor_node] = new_cost

                # f(n) = g(n) + h(n)
                estimated_total_cost = (
                    new_cost + heuristic[neighbor_node]
                )

                heapq.heappush(
                    priority_queue,
                    (
                        estimated_total_cost,
                        new_cost,
                        neighbor_node
                    )
                )

                came_from[neighbor_node] = current_node

    # No path found
    return None, float("inf")
```

```python
start_node = "A"
goal_node = "G"


shortest_path, total_cost = a_star_search(
    graph,
    heuristic,
    start_node,
    goal_node
)


print("Shortest path:", " -> ".join(shortest_path))
print("Total cost:", total_cost)
```
