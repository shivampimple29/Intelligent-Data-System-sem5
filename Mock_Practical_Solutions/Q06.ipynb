# **Question 6: Box Plot and Outlier Identification**

##### **Aim**: To examine the spread of an Iris numerical attribute and identify possible outliers. Draw a box plot for SepalWidthCm and calculate Q1, Q3 and the interquartile range (IQR). Use the 1.5 × IQR rule to identify any possible outlier values and report the result.

###### **Experiment reference**: Experiments 2-3

```python
# Step 1: Import the required libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Step 2: Load the Iris dataset

iris = pd.read_csv('/content/Iris.csv')
df=pd.DataFrame(iris)
print("Shape of dataset : ",df.shape)
print("\n")
display(df.head())
```

```python
# Step 3: Visualize the spread of SepalWidthCm with a box plot

sns.boxplot(x=df['SepalWidthCm'], color='blue')
plt.title('Box Plot of Sepal Width')
plt.xlabel('Sepal Width (cm)')
plt.show()
```

```python
# Step 4: calculate Q1, Q3 and the interquartile range (IQR)

feature=df['SepalWidthCm']
Q1 = feature.quantile(0.25)
q2 = feature.median()
Q3 = feature.quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 *IQR
print(f"Q1: {Q1} | Q2: {Q2} | Q3: {Q3} | IQR: {IQR}")
print()
print(f"Lower Bound: {lower_bound} | Upper Bound: {upper_bound}")
```

```python
# Step 5: Use the 1.5 × IQR rule to identify any possible outlier values and report the result

outliers=df[(feature < lower_bound) | (feature > upper_bound)]

print("Number of possible outliers :",len(outliers))
print("Outlier values              :",sorted(outliers['SepalWidthCm'].tolist()))
print()

display(outliers[['SepalWidthCm','Species']])
print("Boxplot successfully demonstrates 4 outliers and Quartile values and each of those values are verified since they match with numerical results")
```
