# **Question 3: Handling Missing Values**

##### **Aim**: To demonstrate missing-value handling using the Iris dataset. Create a working copy and introduce missing values in a few cells of a numerical column and a categorical column. Detect the missing values, fill numerical gaps with the column median and categorical gaps with the mode, and verify that no missing values remain in those columns.

###### **Experiment reference**: Experiment 2

```python
# Step 1: Import the required libraries

import pandas as pd
import numpy as np
```

```python
# Step 2: Load the Iris dataset

iris = pd.read_csv('/content/Iris.csv')
df=pd.DataFrame(iris)
print("Shape of dataset : ",df.shape)
print("\n")
display(df.head())
```

```python
# Step 3: Work on a copy and insert missing values into two columns

iris_copy=df.copy()
iris_copy.loc[[2, 10, 25], 'SepalLengthCm'] = np.nan
iris_copy.loc[[60,70, 120], 'Species'] = np.nan

print("Missing values before handling:")
print(iris_copy.isnull().sum())

print("\nRows with missing values:\n")
display(iris_copy[iris_copy.isnull().any(axis=1)])
```

```python
# Step 4: Fill numerical gaps with the column median and categorical gaps with the mode

median=iris_copy['SepalLengthCm'].median()
mode=iris_copy['Species'].mode()[0]
print("Median :",median)
print("Mode   :",mode)

iris_copy['SepalLengthCm']=iris_copy['SepalLengthCm'].fillna(median)
iris_copy['Species']=iris_copy['Species'].fillna(mode)

print("\nMissing values after handling:")
print("\n")
display(iris_copy.loc[[2,10,25,60,70,120]])
```

```python
# Step 5:  Verify that no missing values remain in those columns

print("Missing values after handling:")
print(iris_copy.isnull().sum())
```
