# **Question 1: Python Libraries – NumPy**

##### **Aim**: To use NumPy for basic numerical operations on the Iris dataset. Load the Iris data and select a numerical attribute such as SepalLengthCm. Convert its values into a NumPy array and calculate the sum, mean, maximum and minimum. Also display a slice containing the first five values.

###### **Experiment reference**: Experiment 1

```python
# Step 1: Import the required libraries

import pandas as pd
import numpy as np
```

```python
# Step 2: Load the Iris dataset

iris = pd.read_csv('/content/Iris.csv')
df=pd.DataFrame(iris)
print("Shape of dataset : ",df.shape)
print("\n")
print(df.head())
```

```python
# Step 3: Convert SepalLengthCm into a NumPy array

sepal_array=np.array(df['SepalLengthCm'])
print("Array shape : ",sepal_array.shape)
print("Array type  : ",type(sepal_array))
```

```python
#Step 4: Display its sum, mean, maximum, minimum and first five values

print("Sum               : ",np.sum(sepal_array))
print("Mean              : ",np.mean(sepal_array))
print("Maximum           : ",np.max(sepal_array))
print("Minimum           : ",np.min(sepal_array))
print("First five values : ",sepal_array[0:5])

```
