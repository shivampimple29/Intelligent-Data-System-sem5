# **Question 2: Loading and Inspecting the Iris Dataset**

##### **Aim**: To load the Iris dataset into a Pandas DataFrame and inspect its structure. Display the first five records, number of rows and columns, column names, data types and summary statistics. Identify the numerical attributes and the target column, Species.

###### **Experiment reference**: Experiments 1-2

```python
# Step 1: Import the required libraries

import pandas as pd
import numpy as np
```

```python
# Step 2: Load the Iris dataset

iris = pd.read_csv('/content/Iris.csv')
df=pd.DataFrame(iris)
print("Shape of dataset : ",df.shape)
```

```python
# Step 3: Display the first five records, number of rows and columns, column names, data types

print("First five records :\n")
print(df.head())
print("\n")
print("Number of rows    : ",df.shape[0])
print("Number of columns : ",df.shape[1])
print("Number of rows    : ",df.columns.tolist())
print("\nData types        :\n\n",df.dtypes)
```

```python
# Step 4: Display summary statistics & Indentify numerical attributes

print("Summary statistics for numerical attributes :\n")
print(df.describe())
```

```python
# Step 5: Identify the target column & Species

target_col=df['Species']

print("Target Column : ",target_col.name)
print("Species Count :\n\n",target_col.value_counts())
```
