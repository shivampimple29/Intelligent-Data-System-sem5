# **Question 9: Hypothesis Testing – Independent T-test**

##### **Aim**: To determine whether the mean SepalLengthCm differs significantly between Iris-setosa and Iris-versicolor. State the null and alternative hypotheses, perform an independent two-sample t-test using SciPy, and display the t-statistic and p-value. Use a 0.05 significance level to state the conclusion.

###### **Experiment reference**: Experiment 5 (hypothesis testing; adapted to two independent groups)

```python
# Step 1: Import the required libraries

import pandas as pd
from scipy import stats
```

```python
# Step 2: Load the Iris dataset

iris = pd.read_csv('/content/Iris.csv')
df=pd.DataFrame(iris)
print("Shape of dataset : ",df.shape)
print("\n")
display(df.head())
```

```python
# Step 3: State the hypotheses and choose the two independent groups
# H0: mean SepalLengthCm of setosa = mean SepalLengthCm of versicolor
# H1: mean SepalLengthCm of setosa != mean SepalLengthCm of versicolor (two-sided)

setosa=df.loc[df['Species']=='Iris-setosa', 'SepalLengthCm']
versicolor=df.loc[df['Species']=='Iris-versicolor', 'SepalLengthCm']
alpha=0.05
print('H0: μ_setosa = μ_versicolor')
print('H1: μ_setosa != μ_versicolor')
print()
print(f"Setosa : n={len(setosa)}, mean={setosa.mean()}")
print(f"Versicolor : n={len(versicolor)}, mean={versicolor.mean()}")
```

```python
# Step 4: Perform an independent two-sample t-test using SciPy, and display the t-statistic and p-value

t_stat,p_value = stats.ttest_ind(setosa, versicolor, equal_var=False)
print("t-statistic :",t_stat)
print("p-value     :",p_value)
```

```python
#Step 5: Compare p-value with the chosen significance level

print('Significance level (alpha):', alpha)
print()
if p_value < alpha:
    print('Decision: Reject H0.')
    print('Observation: Mean sepal length differs significantly between setosa and versicolor.')
else:
    print('Decision: Fail to reject H0.')
    print('Observation: Not enough evidence of a difference in mean sepal length.')
```
