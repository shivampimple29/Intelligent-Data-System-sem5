# **Question 8: Correlation and Covariance**

##### **Aim**: To study how two numerical attributes of the Iris dataset vary together. Calculate the correlation coefficient and covariance between PetalLengthCm and PetalWidthCm. Display both values and state whether the relationship is positive or negative and whether the correlation appears weak or strong.

###### **Experiment reference**: Experiment 5

```python
# Step 1: Import the required libraries

import pandas as pd
```

```python
# Step 2: Load the Iris dataset

iris = pd.read_csv('/content/Iris.csv')
df=pd.DataFrame(iris)
print("Shape of dataset : ",df.shape)
print("\n")
display(df.head())
```

```python
# Step 3: Calculate the correlation coefficient and covariance between PetalLengthCm and PetalWidthCm

correlation = df['PetalLengthCm'].corr(df['PetalWidthCm'])
covariance  = df['PetalLengthCm'].cov(df['PetalWidthCm'])

print("Correlation :",correlation)
print("Covariance  :",covariance)
```

```python
# Step 4: Interpret the relationship

direction = 'positive' if correlation > 0 else ('negative' if correlation < 0 else 'no linear')
strength = 'strong' if abs(correlation) >= 0.7 else ('moderate' if abs(correlation) >= 0.3 else 'weak')
print(f'Observation: The attributes have a {strength} {direction} linear relationship.')
print('As petal length increases, petal width generally increases.')
```
