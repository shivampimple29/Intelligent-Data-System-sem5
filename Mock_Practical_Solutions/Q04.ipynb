# **Question 4: Data Transformation**

##### **Aim**: To transform the Iris dataset into a form suitable for analysis or machine learning. Convert the Species column into numerical indicator columns using one-hot encoding. Standardize the four numerical attributes and display the transformed data to observe how the values have changed.

###### **Experiment reference**: Experiment 2 (cleaning/transformation approach)

```python
# Step 1: Import the required libraries

import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
```

```python
# Step 2: Load the Iris dataset

iris = pd.read_csv('/content/Iris.csv')
df=pd.DataFrame(iris)
print("Shape of dataset : ",df.shape)
print("\n")
display(df.head())
```

```python
# Step 3: Convert the Species column into numerical indicator columns using one-hot encoding

encoded=pd.get_dummies(df, columns=['Species'], dtype=int)
display(encoded.head())
```

```python
#Step 4: Standardize the four numerical attributes and display the transformed data to observe how the values have changed

scaler=StandardScaler()
cols = ['SepalLengthCm', 'SepalWidthCm', 'PetalLengthCm', 'PetalWidthCm']
encoded[cols] = scaler.fit_transform(encoded[cols])

display(encoded.head())
```

```python
# Step 5: Verify approximate mean 0 and population SD 1

print("Mean of standardized columns :\n")
print(encoded[cols].mean().round(5))

print("Population Standard Deviation :\n")
print(encoded[cols].std(ddof=0))
```
