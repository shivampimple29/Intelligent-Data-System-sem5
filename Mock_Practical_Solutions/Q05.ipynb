##### **Aim**: Perform EDA on the Iris dataset by plotting a histogram of PetalLengthCm and a scatter plot of SepalLengthCm against PetalLengthCm. Add suitable titles, axis labels and different colours for Species in the scatter plot. Briefly state what the plots reveal about the distribution and relationship between the attributes.

###### **Experiment reference**: Experiment 3

```python
# Step 1: Import the required libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Step 2: Load the Iris dataset

iris = pd.read_csv('/content/Iris.csv')
df=pd.DataFrame(iris)
print("Shape of dataset : ",df.shape)
print("\n")
display(df.head())
```

```python
# Step 3: Perform EDA on the Iris dataset by plotting a histogram of PetalLengthCm

sns.histplot(data=df, x='PetalLengthCm', bins=10, color='skyblue')
plt.title('Histogram of PetalLengthCm')
plt.xlabel('PetalLength (cm)')
plt.ylabel('Frequency')
plt.show()
```

```python
# Step 4: EDA on the Iris dataset by plotting a scatter plot of SepalLengthCm against PetalLengthCm

sns.scatterplot(data=df, x='SepalLengthCm', y='PetalLengthCm', hue='Species')
```

```python
# Step 5: A numerical check to support the visual observation

r=df['SepalLengthCm'].corr(df['PetalLengthCm'])
print("Correlation between plotted attrbutes: ",r)
print('Observation: Petal lengths show distinct groups (short setosa petals versus longer petals of the other species).')
print('The scatter plot shows an overall positive association and species-dependent clusters.')
```
