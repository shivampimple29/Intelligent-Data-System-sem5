```python
#Step 1 : Input all the essentials libraries

import numpy as np

arr=[1,2,3,4,5]

#1D array

arr1D=np.array(arr)
print("--------------------------1D Array-------------------------")
print(f"\narray:{arr1D},\ntype:{type(arr1D)},\ndimensions:{arr1D.ndim},\nshape:{arr1D.shape},\nsize:{arr1D.size},\ndatatype:{arr1D.dtype}")

#2D array : (rows,cols)

arr2=[[1,2,3],[4,5,6],[7,8,9]]
arr2D=np.array(arr2)
print("--------------------------2D Array-------------------------")
shape2=arr2D.shape;
rows2=shape2[0]
cols2=shape2[1]
print(f"Number of rows :{rows2}")
print(f"Number of cols :{cols2}")
print(f"\narray:{arr2D},\ntype:{type(arr2D)},\ndimensions:{arr2D.ndim},\nshape:{arr2D.shape},\nsize:{arr2D.size},\ndatatype:{arr2D.dtype}")

#3D array : (depth, rows, cols)

arr3=[[[1,2],[3,4]],[[5,6],[7,8]]]
arr3D=np.array(arr3)
print("--------------------------3D Array-------------------------")
shape3=arr3D.shape;
depth3=shape3[0]
rows3=shape3[1]
cols3=shape3[2]
print(f"Number of depths :{depth3}")
print(f"Number of rows   :{rows3}")
print(f"Number of cols   :{cols3}")
print(f"\narray:{arr3D},\ntype:{type(arr3D)},\ndimensions:{arr3D.ndim},\nshape:{arr3D.shape},\nsize:{arr3D.size},\ndatatype:{arr3D.dtype}")


```

```python
#Array indexing
print("--------------------------Array Indexing-------------------------")
print(f"original array :{arr1D[0:5]}")
print(f"reverse array  :{arr1D[::-1]}")

print(arr2D[0,0]) #first element
print(arr2D[0:3]) #first row

print(arr3D[0,1,1])
print(arr3D[1:2])
```

```python
#Working on Mathematical Functions
print("--------------------------Mathematical Functions-------------------------")
print(f"addition           : {np.sum(arr2D)}")
print(f"average            : {np.mean(arr2D)}")
print(f"minimum            : {np.min(arr2D)}")
print(f"maximum            : {np.max(arr2D)}")
print(f"standard deviation : {np.std(arr2D)}")
print(f"variance           : {np.var(arr2D)}")
```
